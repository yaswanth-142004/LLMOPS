```python
from langchain.document_loaders import PyPDFLoader 
```

```python
file_path = os.path.join(os.getcwd(),"data","sample.pdf")
```

```python
loader = PyPDFLoader(file_path)
```

```python
documents = loader.load()
```

```python
len(documents)
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap = 150,
    length_function = len 
)
```

```python
docs = text_splitter.split_documents(documents)
```

```python
len(docs)
```

```python
from langchain.vectorstores import FAISS 
```

```python
from dotenv import load_dotenv 
load_dotenv()
from langchain_google_genai import GoogleGenerativeAIEmbeddings 
embedding_model = GoogleGenerativeAIEmbeddings(model="models/embedding-001")
doc_vector = embedding_model.embed_query("who is Yaswanth ?")
```

```python
doc_vector
```

```python
len(embedding_model.embed_documents(docs[0].page_content)[0])
```

```python
vector_Store = FAISS.from_documents(docs,embedding_model)
```

```python
relevant_doc = vector_Store.similarity_search("llama2 finetuning benchmark experiemnts")
```

```python
## Keyword Filtering 

retriever = vector_Store.as_retriever(search_kwargs={"k":10})
```

```python
retriever.invoke("llama2 finetuning Benchmark experiments")
```

```python
# Context based on the question retrieveing the info from the vector database 
```

```python
prompt_template = """ 
Answer the queston based on the context provided below . If the context does not contain sufficient information , respond with : "I do not have enough information about this." 

Context :{context}

Question :{question} 

Answer:


"""
```

```python
from langchain.prompts import PromptTemplate 
```

```python
prompt = PromptTemplate(
    template = prompt_template,
    input_variables=["context","question"]
)
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser 
```

```python
parser = StrOutputParser()
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough 
```

```python
from langchain_groq import ChatGroq

lm=ChatGroq(model="deepseek-r1-distill-llama-70b")
```

```python
rag_chain = (
    {"context":retriever|format_docs ,"question" :RunnablePassthrough()}
    |prompt 
    |lm 
    |StrOutputParser()
    
)
```

```python
rag_chain.invoke("Tell me about the llama2 Architecture")
```
